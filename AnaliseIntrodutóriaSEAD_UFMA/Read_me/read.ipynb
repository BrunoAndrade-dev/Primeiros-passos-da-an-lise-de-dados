 ### Trazer uma boa compreenssão da parte introdutória em Análise de Dados :  

- Utilização das bibliotecas Pandas e matplotlib

    - Tratamento de dados com o Pandas

    - Exposição dos resultados com matplotlib


    ###  Relembrando alguns conceitos sobre essas bibliotecas :

    - Pandas :  

        - Carregar dados: Importar dados de vários formatos (CSV, Excel, bancos de dados, etc.) para DataFrames.

        - Tratar dados: Limpar, transformar e organizar seus dados (lidar com valores ausentes, filtrar, selecionar, renomear colunas, etc.).
        
        - Analisar dados: Realizar operações estatísticas, agrupar dados, fazer pivoteamento e muito mais.

    - Matplotlib :  

        - Gráficos de linha: Para mostrar tendências ao longo do tempo.

        - Gráficos de dispersão: Para visualizar a relação entre duas variáveis.

        - Histogramas: Para mostrar a distribuição de uma única variável.
    
        - Gráficos de barras: Para comparar quantidades entre diferentes categorias.

        - Gráficos de pizza: Para mostrar proporções de um todo.
